# Looping over data

Sometimes we need to repeat the same action multiple times on a variable, but don't want to have copy and pasted code.

```python
for number in [2, 3, 5]:
    print(number)
```

Indents determine what is in the loop

```python
for number in [2, 3, 5]:
    print('This number is in the loop:', number)
print('This message is not in the loop')
```
