# Conditionals

```python
mass = 3.54

if mass > 3.0:
    print(mass, 'is large')
```

```python
mass = 2.0

if mass > 3.0:
    print(mass, 'is large')
```

```python
masses = [3.54, 2.07, 9.22, 1.86, 1.71]

for m in masses:
    if m > 9.0:
        print(m, 'is HUGE!!')
    elif m > 3.0:
        print(m, 'is medium')
    else:
        print(m, 'is small')
```

Conditions are tested once, in order

```python
grade = 85

if grade >= 90:
    print('grade is A')
elif grade >= 80:
    print('grade is B')
elif grade >= 70:
    print('grade is C')
```

```python
grade = 85

if grade >= 90:
    print('grade is A')
if grade >= 80:
    print('grade is B')
if grade >= 70:
    print('grade is C')
```

```python
grade = 85

if grade >= 70:
    print('grade is C')
elif grade >= 80:
    print('grade is B')
elif grade >= 90:
    print('grade is A')
```

If statements do not reevaluate if changes are made

```python
velocity = 10.0

if velocity > 20.0:
    print('moving too fast')
else:
    print('adjusting velocity')
    velocity = 50.0
```
